```python
num = int(input('정수를 입력하세요: '))

if num % 2 == 0:
    print(f'{num}은(는) 짝수입니다.')
else:
    print(f'{num}은(는) 홀수입니다.')
```

```python
import random

# 1: rkdnl, 2: 바위, 3: 보
com = random.randint(1, 3)
user = int(input('가위(1), 바위(2), 보(3) 중 선택: '))

if com == 1:
  print('컴퓨터: 가위')
elif com == 2:
  print('컴퓨터: 바위')
else:
  print('컴푸터: 보')

if user == com:
   print('비겼습니다.')
elif user == 1 and com == 3:
  print('이겼습니다!')
elif user == 2 and com == 1:
  print('이겼습니다!')
elif user == 3 and com == 2:
  print('이겼습니다!')
else:('졌습니다.')
```

```python
import random
com = random.randint(1, 100)
count=0
while True:
    user = int(input("당신이 생각하는 수"))
    count+=1
    print(com)
    if user == com:
        print('정답입니다..')
        print("시도한 횟수: ",count)
        break
    elif user>com:
        print('입력한 숫자가 큽니다.')
    else:
        print('입력한 숫자가 작습니다..')
```

```python
def asmd(x,y,op):
    if op=='+':

        print(f'{x} + {y} = {x, y}')
    elif op=='-':
        print(f'{x} - {y} = {x, y}')
     elif op=='*':
        print(f'{x} * {y} = {x, y}') 
     elif op=='/':
        print(f'{x} / {y} = {x, y}')
    else:
        print("잘못된 연산자입니다.")
        
        
```

```python
x = int(input('첫 번째 수: '))
y = int(input('두 번째 수: '))
op=input('연산자를 입력하시오')
asmd(x,y,op)
```
